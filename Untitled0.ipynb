```python

scores = [80, 65, 90, 75, 70, 85]

# Soal 1
print("Nilai pertama:", scores[0])
print("Nilai terakhir:", scores[-1])

# Soal 2
rata_rata = sum(scores) / len(scores)
print("Rata-rata:", rata_rata)

jumlah = 0
for nilai in scores:
    if nilai > rata_rata:
        jumlah += 1

print("Jumlah nilai di atas rata-rata:", jumlah)

# Soal 3
nilai_terendah = min(scores)
posisi = scores.index(nilai_terendah) + 1

print("Nilai terendah:", nilai_terendah)
print("Posisi siswa:", posisi)

# Soal 4
nilai_baru = []

for nilai in scores:
    nilai_baru.append(nilai + 5)

print("Nilai setelah ditambah 5:", nilai_baru)
```
