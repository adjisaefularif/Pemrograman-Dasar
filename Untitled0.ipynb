```python
tekanan_uap = float(input("Masukkan tekanan uap (psi): "))

status_manual_override = False

if status_manual_override == True:
    print("Katup Mode Manual")
elif tekanan_uap > 150 and status_manual_override == False:
    print("Katup Terbuka")
else:
    print("Katup Tertutup")
```

Penjelasan:
Program ini digunakan untuk menentukan kondisi katup pengaman berdasarkan tekanan uap dan status manual. Nilai tekanan dimasukkan melalui keyboard menggunakan input(). Percabangan if-elif-else digunakan untuk menentukan apakah katup terbuka, dalam mode manual, atau tertutup. Jika mode manual aktif, maka katup berada dalam mode manual.

```python
suhu_saat_ini = 60
target = 85

while suhu_saat_ini < target:
    suhu_saat_ini += 5

    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print(f"Target {target} C tercapai.")
```

Penjelasan:
Program ini digunakan untuk menyimulasikan kenaikan suhu pada sistem pemanas. Suhu dimulai dari 60°C dan dinaikkan 5°C setiap perulangan menggunakan while. Ketika suhu mencapai 80°C, program menampilkan peringatan. Perulangan akan berhenti setelah suhu mencapai 85°C.

```python
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

jumlah_tidak_valid = 0

for nomor, nilai in enumerate(data, start=1):
    try:
        vibrasi = float(nilai)

    except ValueError:
        print(f"Unit {nomor}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    if vibrasi > 2.0:
        print(f"Warning: Unit {nomor} vibrasi tinggi!")

    if vibrasi > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

print("Jumlah data tidak valid:", jumlah_tidak_valid)
```

Penjelasan:
Program ini digunakan untuk memeriksa data vibrasi dari lima unit mesin. Data disimpan dalam bentuk list dan diperiksa satu per satu menggunakan for. try-except digunakan untuk menangani data yang tidak bisa diubah menjadi angka, seperti ERR. Jika data tidak valid, program melewatinya menggunakan continue. Jika vibrasi lebih dari 2.0, muncul peringatan. Jika lebih dari 3.0, program menghentikan pemeriksaan menggunakan break.

```python
# Data beban listrik selama 5 hari (kW)
beban_listrik = [350, 420, 480, 550, 390]

batas_beban = 500
total_beban = 0

print("=== PEMANTAUAN BEBAN LISTRIK ===")

for hari, beban in enumerate(beban_listrik, start=1):

    total_beban += beban

    print(f"Hari ke-{hari}: {beban} kW")

    if beban > batas_beban:
        print("STATUS: BEBAN BERLEBIH")

    elif beban >= 400:
        print("STATUS: BEBAN MENINGKAT")

    else:
        print("STATUS: NORMAL")

rata_rata = total_beban / len(beban_listrik)

print("-----------------------------")
print("Total beban listrik:", total_beban, "kW")
print(f"Rata-rata beban: {rata_rata:.2f} kW")

if rata_rata > batas_beban:
    print("KESIMPULAN: Beban listrik tinggi")

elif rata_rata >= 400:
    print("KESIMPULAN: Beban listrik cukup tinggi")

else:
    print("KESIMPULAN: Beban listrik normal")
```

Penjelasan:
Program ini dibuat untuk memantau beban listrik selama lima hari. Data beban disimpan dalam list, kemudian diperiksa menggunakan perulangan for. Percabangan if-elif-else digunakan untuk menentukan kondisi beban listrik, apakah normal, meningkat, atau berlebihan. Selanjutnya program menghitung total dan rata-rata beban listrik untuk mengetahui kondisi penggunaan listrik selama lima hari.
