```python
# Jawaban Latihan 4
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir
print("Data pertama:", tegangan_harian[0])
print("Data terakhir:", tegangan_harian[-1])

# (b) Tambahkan data baru
tegangan_harian.append(222.0)
print("List setelah ditambah:", tegangan_harian)

# (c) Cetak jumlah total data
print("Jumlah total data:", len(tegangan_harian))
```

```python
# Jawaban Latihan 5
motor_listrik = {
    "nama": "Motor Induksi 3-Fasa",
    "daya_kW": 15,
    "tegangan_V": 380,
    "rpm": 1450
}

# Cetak salah satu nilai lewat key-nya (misal: daya_kW)
print("Daya motor:", motor_listrik["daya_kW"], "kW")

# Ubah nilai rpm
motor_listrik["rpm"] = 1500

# Cetak ulang dictionary
print("Data motor diperbarui:", motor_listrik)
```

```python
# Jawaban Latihan 6
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

# 1. Cetak nama unit kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

# 2. Ubah status unit ke-3 (indeks 2)
pembangkit[2]["status"] = "operasi"

# 3. Tambahkan unit baru (Unit-4) dengan append()
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 120, "status": "operasi"}
pembangkit.append(unit_baru)

# 4. Cetak seluruh data
print("\nSeluruh Data Pembangkit:")
print(pembangkit)
```
