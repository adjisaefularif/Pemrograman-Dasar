```python
# Soal 1: Belanja Buku

jumlah_buku = int(input("Jumlah buku: "))
harga_buku = int(input("Harga per buku: "))
diskon_persen = float(input("Diskon (%): "))
uang_budi = int(input("Uang Budi: "))
jumlah_orang = int(input("Jumlah orang: "))

total = jumlah_buku * harga_buku
diskon = int(total * diskon_persen / 100)
bayar = total - diskon
kembalian = uang_budi - bayar

per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang
uang_cukup = uang_budi >= bayar

print("\nTotal       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", uang_cukup)
```

```python
# Soal 2: PLTS Atap Rumah

jumlah_panel = int(input("Jumlah panel: "))
daya_panel = int(input("Daya per panel (Wp): "))
jam_matahari = float(input("Penyinaran efektif (jam): "))
efisiensi = float(input("Efisiensi sistem (contoh 0.8): "))
kebutuhan = float(input("Kebutuhan listrik (kWh/hari): "))
tarif = float(input("Tarif listrik per kWh (Rp): "))

energi_harian = (
    jumlah_panel * daya_panel * jam_matahari
    * efisiensi / 1000
)

energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian

if kekurangan < 0:
    kekurangan = 0

mencukupi = energi_harian >= kebutuhan

print("\nEnergi harian (kWh) :", round(energi_harian, 2))
print("Energi bulanan (kWh):", round(energi_bulanan, 2))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 2))
print("Kekurangan (kWh)    :", round(kekurangan, 2))
print("Mencukupi?          :", mencukupi)
```

```python
# Soal 3: Generator Diesel dan Rugi Kabel

daya_generator = float(input("Daya generator (kW): "))
jam_operasi = float(input("Jam operasi per hari: "))
konsumsi_solar = float(input("Konsumsi solar (liter/kWh): "))
harga_solar = float(input("Harga solar per liter (Rp): "))
kapasitas_tangki = int(input("Kapasitas tangki (liter): "))
arus = float(input("Arus kabel (A): "))
hambatan = float(input("Hambatan kabel (Ohm): "))

energi = daya_generator * jam_operasi
solar = energi * konsumsi_solar
biaya = solar * harga_solar

hari_penuh = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar

rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("\nEnergi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", f"{hari_penuh} hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
